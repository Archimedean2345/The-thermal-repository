```python
import numpy as np

E = 200e9
A1, L1 = 2e-4, 0.5   # Barra 1
A2, L2 = 1e-4, 0.3   # Barra 2 (más delgada y corta, a propósito, para que no sean iguales)

k1 = E*A1/L1
k2 = E*A2/L2

# Matriz global 3x3
K = np.zeros((3,3))

# Ensamblaje: barra 1 afecta nodos 0 y 1
K[0,0] += k1
K[0,1] += -k1
K[1,0] += -k1
K[1,1] += k1

# Ensamblaje: barra 2 afecta nodos 1 y 2
K[1,1] += k2
K[1,2] += -k2
K[2,1] += -k2
K[2,2] += k2

F = np.array([0, 0, 5000])   # Fuerza solo en nodo 2

# Nodo 0 fijo → resolvemos solo para nodos 1 y 2
u = np.linalg.solve(K[1:,1:], F[1:])
print(u)
```

```python
# Desplazamientos completos (incluyendo nodo 0 fijo)
u_full = np.array([0, [0], u[1]])   # [u_nodo0, u_nodo1, u_nodo2]

# Strain en cada barra = (u_nodo_derecha - u_nodo_izquierda) / L
strain_barra1 = (u_full[1] - u_full[0]) / L1
strain_barra2 = (u_full[2] - u_full[1]) / L2

# Stress en cada barra = E * strain
stress_barra1 = E * strain_barra1
stress_barra2 = E * strain_barra2

print("Strain:", strain_barra1, strain_barra2)
print("Stress (Pa):", stress_barra1, stress_barra2)
print("Fuerza en cada barra (N):", stress_barra1*A1, stress_barra2*A2)
```
